### Deliverable 3
Presentation of each class & battle simulations

# Warrior class

A defensive specialist subclass inheriting from the base Character class.

Attributes:
name (str), _hp (int), hp_max (int), damage (int), armor (int) weapon (None).

Methods:
__init__(self, name: str, hp_max: int = 100, damage: int = 10, armor:: int = 5,  weapon: Weapon = None) -> None
: Calls super().__init__() and initializes self.armor.

take_damage(self, damage: int) -> None (Overridden): Overrides Character.take_damage() by subtracting armor from incoming damage (effective_damage = max(0, damage - self.armor) ) before passing the reduced damage to super().take_damage()

# Mage class

A resource-driven attacker subclass inheriting from the base Character class.

Attributes:
name (str), _hp (int), hp_max (int), damage (int), weapon (None), mana (int), mana_max (int).

Methods:
__init__(self, name: str, hp_max: int = 100, damage: int = 10, mana: int = 30, mana_max: int = 30, weapon: Weapon = None) -> None
: Calls super().__init__() and initializes self.mana and self.mana_max.

attack(self, target: Character) -> None (Overridden):
if self.mana >= 10: Consumes 10 mana to cast a spell dealing double base damage (self.damage * 2) directly to the target without rolling weapon damage.
if self.mana < 10: Falls back to standard base attack via super().attack(target) (which includes weapon damage if equipped) and restores 5 mana up to mana_max.

# Battle class

Manages turn-based combat between two Character objects.

Attributes:
hero (Character): The first combatant.
enemy (Character): The second combatant.

Methods:
__init__(self, hero: Character, enemy: Character) -> None
: Connects the two combatants to the battle instance.

play_round(self) -> None: Executes a single combat turn: hero attacks enemy first, and if the enemy.is_alive is True, enemy retaliates against hero.

simulate_battle(self) -> Character: Executes play_round() in a while loop until one character reaches 0 HP (while self.hero.is_alive and self.enemy.is_alive), then returns the winning Character instance.
