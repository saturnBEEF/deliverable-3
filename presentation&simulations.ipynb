# Warrior class

A defensive specialist subclass inheriting from the base Character class.

Attributes:
name (str), hp_max (int), damage (int), armor (int) weapon (None).

Methods:
__init__(self, name: str, hp_max: int = 100, damage: int = 10, armor:: int = 5,  weapon: Weapon = None) -> None
: Calls super().__init__() and initializes self.armor.

take_damage(self, damage: int) -> None (Overridden): Overrides Character.take_damage() by subtracting armor from incoming damage (effective_damage = max())
